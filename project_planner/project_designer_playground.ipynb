```python
project_template = """\
Name:
AutoMentor: Autonomous AI Learning Assistant

Objectives:

To develop an agentic AI-powered virtual learning assistant that:
* Autonomously plans and adjusts personalized learning paths
* Interacts with students via natural language and visual aids
* Proactively recommends resources, quizzes, and explains mistakes
* Operates within guardrails set by educators and curriculum constraints

Budget Constraints:

* Total project budget: $180,000
* 45% allocated to AI agent development (LLM fine-tuning, RAG system)
* 25% to platform backend and integrations (LMS APIs, database)
* 20% to front-end and interaction design (chat UI, voice/vision)
* 10% for testing, compliance, and contingency

Timeline (Interrelated with Resources):

* Project duration: 4 months
* Major milestones:
    * Month 1: Scope definition, system design, initial dataset preparation
    * Month 2: Agent core logic implementation, backend scaffolding
    * Month 3: Integration with LMS and UI prototyping, user feedback loop
    * Month 4: Evaluation, classroom trials, educator onboarding, deployment

Sensitivity Requirements:

* Educational integrity: AI must not complete assignments for students
* Transparency: Students and teachers must understand agent behavior (explainability layer required)
* Compliance: FERPA-compliant data handling
* Safety: Must include reinforcement learning with human feedback (RLHF) to align with institutional goals
"""
```

```python
from semantic_kernel.contents import FunctionCallContent, FunctionResultContent
from semantic_kernel.contents.chat_message_content import ChatMessageContent
from IPython.display import display, Markdown

async def handle_intermediate_steps(message: ChatMessageContent) -> None:
    for item in message.items or []:
        if isinstance(item, FunctionCallContent):
            display(Markdown(f"FUNCTION_CALL :> {item.name} with arguments: {item.arguments}"))
        elif isinstance(item, FunctionResultContent):
            display(Markdown(f"FUNCTION_RESULT :> {item.result} for function: {item.name}"))
        else:
            display(Markdown(print(f"LLM :> {message.role}: {message.content}")))

async for response in master_agent.invoke(
            messages=project_template,
            on_intermediate_message=handle_intermediate_steps,
        ):
            display(Markdown(f"# {response.role}: {response}"))
```
